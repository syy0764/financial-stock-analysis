```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from pathlib import Path

Path.cwd()
```

```python
base_rate_summary = pd.read_csv("../data/processed/base_rate_summary.csv")

base_rate_summary.head(5)
```

```python
print(base_rate_summary.shape)
base_rate_summary.info()
base_rate_summary.isna().sum()
```

```python
kb_summary = base_rate_summary[base_rate_summary["bank_name"] == "KB금융"].sort_values("base_rate")

print(kb_summary)
```

```python
base_rate_summary["bank_name"] == "KB금융"
```

```python
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
kb_summary.plot(kind = "bar", x = "base_rate", 
                y = ["avg_bank_return", "avg_kospi_return"],
                figsize=(10,5), rot = 45,
                title = "기준금리 수준별 KB금융과 KOSPI 평균 일별 수익률",
                xlabel = "기준금리(%)",
                ylabel = "평균 일별 수익률 (%)")

plt.legend(["KB금융", "KOSPI"])
plt.savefig("../results/kb_base_rate_return.png", dpi=150, bbox_inches="tight")
plt.show()
```

기준금리가 높아질수록 KB금융의 평균 일별 수익률이 일관되게 증가하는 모습은 나타나지 않았다. 
금리 구간마다 거래일 수와 당시 시장 상황이 다르기 때문에 수익률 차이를 금리 자체의 효과로 단정할 수 없다.

```python
shinhan_summary = base_rate_summary[
    base_rate_summary["bank_name"] == "신한지주"
].sort_values("base_rate")
```

```python
shinhan_summary.plot(kind="bar", x="base_rate", 
                     y = ["avg_bank_return", "avg_kospi_return"], 
                     figsize = (10, 5), rot = 45,
                title = "기준금리 수준별 신한지주와 KOSPI 평균 일별 수익률",
                xlabel = "기준금리(%)",
                ylabel = "평균 일별 수익률 (%)")

plt.legend(["신한지주", "KOSPI"])
plt.savefig("../results/shinhan_base_rate_return.png", dpi=150, bbox_inches="tight")
plt.show()
```
