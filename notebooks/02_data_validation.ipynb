```python
market_files = {
    "KB금융": "kb_financial.csv",
    "신한지주": "shinhan_financial.csv",
    "하나금융지주": "hana_financial.csv",
    "우리금융지주": "woori_financial.csv",
    "KOSPI": "kospi.csv",
    "USD_KRW": "usd_krw.csv"
}
```

```python
market_data_raw = {}

for name, file_name in market_files.items():
    data = pd.read_csv(raw_dir / file_name)
    market_data_raw[name] = data
    print(name, data.shape)
```

```python
base_rate_raw = pd.read_csv(raw_dir / "bok_base_rate.csv")
base_rate_raw.head(5)
```

```python
print(base_rate_raw.shape)
print(base_rate_raw.dtypes)
```

```python
for name, data in market_data_raw.items():
    if name == "USD_KRW":
        date_col = "Unnamed: 0"
    else:
        date_col = "Date"

    print(name)
    print(data.columns)
    print("시작일:", data[date_col].min())
    print("종료일:", data[date_col].max())
    print()
```

```python
for name, data in market_data_raw.items():
    if name == "USD_KRW":
        date_col = "Unnamed: 0"
    else:
        date_col = "Date"

    duplicate_count = data[date_col].duplicated().sum()
    print(name, duplicate_count)
```

```python
print(base_rate_raw.isna().sum())
print(base_rate_raw["change_date"].duplicated().sum())
```

```python
for name, data in market_data_raw.items():
    invalid_close = (data["Close"] <= 0).sum()
    invalid_high_low = (data["High"] < data["Low"]).sum()
    negative_volume = (data["Volume"] < 0).sum()

    print(name, invalid_close, invalid_high_low, negative_volume)
```

```python
start_date = base_rate_raw["change_date"].min()
end_date = base_rate_raw["change_date"].max()

min_rate = base_rate_raw["base_rate"].min()
max_rate = base_rate_raw["base_rate"].max()
negative_rate_count = (base_rate_raw["base_rate"] < 0).sum()

print(start_date, end_date)
print(min_rate, max_rate, negative_rate_count)
```

```python
fred_raw = pd.read_csv(
    raw_dir / "usd_krw_fred.csv",
    index_col=0,
    parse_dates=True,
    na_values="."
)

print("전체 행 수:", len(fred_raw))
print("시작일:", fred_raw.index.min())
print("종료일:", fred_raw.index.max())
print("중복 날짜 수:", fred_raw.index.duplicated().sum())
print("환율 결측치 수:", fred_raw["DEXKOUS"].isna().sum())
print("0 이하 환율 수:", (fred_raw["DEXKOUS"] <= 0).sum())
```

## 원본 데이터 검증 결과

### 1. 데이터 범위

- 금융주 4개 및 KOSPI: 2021-01-04 ~ 2025-12-30
- USD/KRW: 2020-12-31 ~ 2025-12-30
- 한국은행 기준금리: 2020-05-28 ~ 2025-05-29
- 최종 분석용 FRED 환율: 2021-01-04 ~ 2025-12-31, 1,303행

### 2. 결측치 및 중복

- 금융주 4개 및 KOSPI: 결측치 없음, 중복 날짜 없음
- 한국은행 기준금리: 결측치 없음, 중복 변경일 없음
- USD/KRW: 중복 날짜 없음, 숫자 열 전체가 비어 있는 4행 확인
  - 2025-01-01
  - 2025-04-17
  - 2025-04-20
  - 2025-12-25
- 최종 분석용 FRED 환율: 환율 결측치 54개, 중복 날짜 없음. 결측값은 유지함

### 3. 구조상 확인 사항

- 시장 데이터의 날짜 열은 CSV를 다시 불러온 뒤 문자열형으로 확인됨
- USD/KRW의 날짜 열 이름은 `Unnamed: 0`으로 저장됨
- USD/KRW에 분석 시작일 이전인 2020-12-31 자료 1행이 포함됨
- 기준금리의 2020-05-28 자료는 2021-01-01 당시 적용 금리인 0.50%를 반영하기 위해 유지함

### 4. 값 범위 검사

- 시장 데이터의 0 이하 종가: 없음
- 고가가 저가보다 작은 행: 없음
- 음수 거래량: 없음
- 기준금리 범위: 0.50% ~ 3.50%
- 음수 기준금리: 없음
- 최종 분석용 FRED 환율의 0 이하 값: 없음

### 5. SQL 정제 기준

- 원본 CSV는 수정하지 않는다.
- 날짜와 숫자 열을 SQL에서 분석에 필요한 자료형으로 변환한다.
- 최종 환율 분석에는 FRED DEXKOUS 자료를 사용한다.
- FRED 환율의 결측값은 삭제하거나 이전 값으로 채우지 않고 유지한다.
- 각 은행의 거래일을 기준으로 KOSPI와 FRED 환율을 같은 날짜로 결합한다.
- 기준금리는 해당 거래일까지 발표된 변경일 중 가장 최근 날짜의 금리를 연결한다.
- 환율 변화율은 KB금융 거래일을 기준으로 계산한다.
- 환율 관계 분석에는 은행 수익률과 환율 변화율이 모두 있는 날짜만 사용한다.
